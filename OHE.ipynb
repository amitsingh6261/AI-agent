```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# 1. Create data
df = pd.DataFrame({
    "Name": ["Amit", "Rahul", "Priya", "Ravi", "Neha"],
    "City": ["Indore", "Delhi", "Mumbai", "Indore", "Delhi"],
    "Gender": ["Male", "Male", "Female", "Male", "Female"]
})

print("Original Data:")
print(df)

```

```python
ohe = OneHotEncoder(sparse_output=False)
```

```python
encoded = ohe.fit_transform(df[["City", "Gender"]])
```

```python
columns = ohe.get_feature_names_out(["City", "Gender"])
```

```python
encoded_df = pd.DataFrame(encoded, columns=columns)

print("\nEncoded Data:")
print(encoded_df)
```
